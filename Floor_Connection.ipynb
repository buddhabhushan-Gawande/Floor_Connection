```python
import pandas as pd

def extract_columns_from_excel(input_file, output_file):
    # Load the Excel file with all sheets
    xls = pd.ExcelFile(input_file)

    # Define the columns to extract (1, 2, 5, 6, 8 - zero-indexed)
    cols_to_extract = [0, 1, 4, 5, 7]

    # Create an empty list to store dataframes from each sheet
    all_sheets_data = []

    # Loop through all sheets
    for sheet_name in xls.sheet_names:
        df = pd.read_excel(xls, sheet_name=sheet_name, usecols=cols_to_extract)
        # Add a sheet name column for identification
        df['Sheet Name'] = sheet_name
        all_sheets_data.append(df)

    # Concatenate all dataframes into a single dataframe
    final_df = pd.concat(all_sheets_data, ignore_index=True)

    # Write the extracted data to a new Excel file
    final_df.to_excel(output_file, index=False)

# Example usage
input_file = "/content/SCREEN HOUSE (MOIL).xlsx"
output_file = "extracted_data.xlsx"
extract_columns_from_excel(input_file, output_file)
```
